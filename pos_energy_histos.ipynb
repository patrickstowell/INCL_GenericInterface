```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the data from your .txt file
# Replace 'data.txt' with the actual path to your file
data = np.loadtxt('pos_energy.txt')

# Separate the data into two 1D arrays
col1 = data[:, 0]
col2 = data[:, 1]
```

```python
# ---------------------------------------------------------
# 2. Create the two 1D histograms
# ---------------------------------------------------------
plt.figure(figsize=(12, 5))

# First 1D Histogram (Column 1)
plt.subplot(1, 2, 1) # 1 row, 2 columns, 1st plot
plt.hist(col1, bins=50, color='skyblue', edgecolor='black')
plt.title('Position distribution Pre swap')
plt.xlabel('Position')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Second 1D Histogram (Column 2)
plt.subplot(1, 2, 2) # 1 row, 2 columns, 2nd plot
plt.hist(col2, bins=50, color='skyblue', edgecolor='black')
plt.title('Pre swap Energy Distribution')
plt.xlabel('Energy')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# ---------------------------------------------------------
# 3. Create the 2D histogram
# ---------------------------------------------------------
plt.figure(figsize=(12, 6))

# hist2d takes the x data, y data, and number of bins
plt.hist2d(col1, col2, bins=30, cmap='plasma')

# Add a colorbar to show the scale of the counts
plt.colorbar(label='Counts')
plt.title('SF pos/mom')
plt.xlabel('Position [fm]')
plt.ylabel('Momentum [MeV/c]')
plt.ylim(0,1000)
plt.xlim(0,10)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the data from your .txt file
# Replace 'data.txt' with the actual path to your file
data = np.loadtxt('pos_energy_postswap.txt')

# Separate the data into two 1D arrays
col3 = data[:, 0]
col4 = data[:, 1]
```

```python
# ---------------------------------------------------------
# 3. Create the 2D histogram
# ---------------------------------------------------------
plt.figure(figsize=(12, 6))

# hist2d takes the x data, y data, and number of bins
plt.hist2d(col4, col3, bins=30, cmap='plasma')

# Add a colorbar to show the scale of the counts
plt.colorbar(label='Counts')
plt.title('incl pos/mom')
plt.xlabel('Position [fm]')
plt.ylabel('Momentum [MeV/c]')
plt.ylim(0,1000)
plt.xlim(0,10)

plt.show()
```

```python
# ---------------------------------------------------------
# 2. Create the two 1D histograms
# ---------------------------------------------------------
plt.hist([col1, col4], bins=30, histtype='step', color=['green', 'blue'], 
         label=['NEUT', 'INCL'], linewidth=1.5)

plt.title('struck neutron position distribution')
plt.xlabel('Position')
plt.xlim(0,10)
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python

# Use histtype='step' for unfilled outlines, and pass colors as a list
plt.hist([col2, col3], bins=100, histtype='step', color=['green', 'blue'], 
         label=['Pre Scatter', 'Post Scatter'], linewidth=1.5)

plt.title('scattered neutron Momentum distribution')
plt.xlabel('MOmentum [MeV/c]')
plt.ylabel('Frequency')
plt.legend()

#plt.tight_layout()
plt.show()
```

```python
plt.subplot(1, 2, 2) # 1 row, 2 columns, 2nd plot
plt.hist(col3-col2, bins=50, color='lightgreen', edgecolor='black')
plt.title('1D Histogram - Column 2')
plt.xlabel('Momentum Difference (INCL-NEUT)')
plt.ylabel('Frequency')

mean_val = (col3 - col2).mean()

# To display it on your plot (optional but helpful)
plt.axvline(mean_val, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean_val:.2f}')
plt.legend()
plt.tight_layout()
plt.show()
```
