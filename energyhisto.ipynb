```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
counts, bin_edges, _ = plt.hist(df['Values'], bins=500, color='skyblue', 
                                edgecolor='black', alpha=0.7, density=True, histtype="step", label='Data')


bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

max_index = np.argmax(counts)
peak_x = bin_centers[max_index]
peak_y = counts[max_index]


i=0
for i in range(max_index):
    #bin_centers[max_index+i] = bin_centers[max_index-i]
    counts[max_index+i] = counts[max_index-i]

plt.plot(bin_centers[:2*max_index], counts[:2*max_index], color='red', linewidth=2)
plt.title('Remnant excitation energy', fontsize=15)
plt.xlabel('Excitation Energy [MeV]', fontsize=12)
plt.xlim(0, 50)
plt.ylim(0, 0.01)
plt.ylim(1e-4, 0.1)
# Updated label to reflect normalization
plt.ylabel('Probability Density', fontsize=12) 
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

```python
bin_centers2 = bin_centers[:2*max_index]
count2       = counts[:2*max_index]

initial_guess = [peak_y, peak_x, 1.0]
popt, pcov = curve_fit(gaussian, bin_centers2, counts2, p0=initial_guess)

print(f"Fitted Mean: {popt[1]:.3f} MeV")
print(f"Fitted StdDev: {popt[2]:.3f} MeV")

# 6. Generate the fitted curve line
x_fit = np.linspace(0, 50, 1000) # Smooth x-axis for plotting the curve
y_fit = gaussian(x_fit, *popt)   # Unpack the fitted parameters into the function

# Plot the Gaussian fit
plt.plot(x_fit, y_fit, color='red', linewidth=2, label=f'Gaussian Fit ($\mu={popt[1]:.2f}$)')

# 7. Add labels and styling (incorporating your original code)
plt.title('Remnant excitation energy', fontsize=15)
plt.xlabel('Excitation Energy [MeV]', fontsize=12)
plt.xlim(0, 50)
plt.ylim(1e-4, 0.1) # Note: You had two ylims in your prompt, I kept the final one
plt.ylabel('Probability Density', fontsize=12) 
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()

# Show the plot
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv('e_star_srcon_srcoff.txt', sep='\s+', header=None, names=['Column1', 'Column2'])

# Filter data as per your requirements
y1 = df.loc[df['Column1'] > -1000, 'Column1'].to_numpy()
y2 = df.loc[df['Column2'] > -1000, 'Column2'].to_numpy()

# Define identical bins
shared_bins = np.linspace(-150, 150, 501)

# 1. Calculate the bin counts for each histogram
counts1, _ = np.histogram(y1, bins=shared_bins)
counts2, _ = np.histogram(y2, bins=shared_bins)

# 2. Subtract the histograms (Column 2 - Column 1)
subtracted_counts = counts2 - counts1

plt.figure(figsize=(10, 6))

# Plot Histogram 2 (Original y2 / Source Off)
#plt.hist(y1, bins=shared_bins, alpha=0.5, label='SRC Contribution', histtype="step", color='lightcoral')


# Plot the Subtracted Histogram
# We use the bin edges and pass the subtracted_counts as weights
plt.hist(shared_bins[:-1], bins=shared_bins, weights=subtracted_counts, 
         alpha=0.8, label='Subtracted (Off - On)', histtype="step", color='skyblue', linewidth=1.5)

# Formatting
plt.yscale("log")
plt.xlim(-150, 150)
plt.xlabel('Energy (or relevant unit)')
plt.ylabel('Counts')
plt.title('Source-Off Histogram vs. Subtracted Histogram (Off - On)')
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the data
file_path = 'missing_E_aftermerge.txt' 
df = pd.read_csv(file_path, header=None, names=['Values'])

# Determine bin edges so that each bin is exactly 0.5 MeV wide
# We calculate the edges from the minimum to the maximum value in the dataset
min_val = df['Values'].min()
max_val = df['Values'].max()
bin_edges = np.arange(min_val, max_val + 0.25, 0.25)

# 2. Create the plot
plt.figure(figsize=(10, 6))

# Set density=True to normalise the bins, use the calculated bin_edges
plt.hist(df['Values'], bins=bin_edges, color='skyblue', edgecolor='red', alpha=0.7, density=True, histtype="step")

# 3. Add labels and styling
plt.title('Missing Energy, before FSI', fontsize=15)
plt.xlabel('Missing E [MeV]', fontsize=12)
plt.xlim(-1, 50)
plt.ylim(1e-4, 1)
#plt.yscale('log')
# Updated label to reflect normalization
plt.ylabel('Probability Density', fontsize=12) 
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Initialize lists to store the values for the histograms
single_values = []
second_values = []

# Open and read the text file
with open('missing_E_Estar_aftermerge.txt', 'r') as file:
    for line in file:
        parts = line.strip().split()
        if len(parts) == 1:
            single_values.append(float(parts[0]))
        elif len(parts) == 2:
            second_values.append(float(parts[1]))

# Create the figure and subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

bin_width = 0.25

min_val1 = min(single_values)
max_val1 = max(single_values)
bins_list1 = np.arange(min_val1, max_val1 + bin_width, bin_width)

ax1.hist(single_values, bins=bins_list1, color='skyblue', edgecolor='none',density="true")
ax1.set_title('Missing Energy Distribution (no recoil)')
ax1.set_xlim(0,50)
ax1.set_xlabel('E_m [MeV]')
ax1.set_ylabel('Frequency')
ax1.axvline(x=18.7, color='red', linestyle='dashed', linewidth=2, label='18.4 MeV')
ax1.legend() # Show the legend to display the label

min_val2 = min(second_values)
max_val2 = max(second_values)
bins_list2 = np.arange(min_val2, max_val2 + bin_width, bin_width)

ax2.hist(second_values, bins=bins_list2, color='salmon', edgecolor='coral', histtype="step", density="true")
ax2.set_title('Excitation Energy of the Remnant')
ax2.set_xlim(-5,25)
ax2.set_ylim(10**-3,1)
ax2.set_xlabel('E* [MeV]')
ax2.set_ylabel('Frequency')
ax2.set_yscale("log")
ax2.axvline(x=0, color='blue', linestyle='dashed', linewidth=2, label='0.0 MeV')
ax2.legend() # Show the legend to display the label


plt.tight_layout()
plt.show()
```
