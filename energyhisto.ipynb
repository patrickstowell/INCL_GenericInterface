```python
def gaussian(x, amplitude, mean, stddev):
    return amplitude * np.exp(-((x - mean) / stddev)**2 / 2)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
counts, bin_edges, _ = plt.hist(df['Values'], bins=500, color='skyblue', 
                                edgecolor='black', alpha=0.7, density=True, histtype="step", label='Data')


bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

max_index = np.argmax(counts)
peak_x = bin_centers[max_index]
peak_y = counts[max_index]


i=0
for i in range(max_index):
    #bin_centers[max_index+i] = bin_centers[max_index-i]
    counts[max_index+i] = counts[max_index-i]

plt.plot(bin_centers[:2*max_index], counts[:2*max_index], color='red', linewidth=2)
plt.title('Remnant excitation energy', fontsize=15)
plt.xlabel('Excitation Energy [MeV]', fontsize=12)
plt.xlim(0, 50)
plt.ylim(0, 0.01)
plt.ylim(1e-4, 0.1)
# Updated label to reflect normalization
plt.ylabel('Probability Density', fontsize=12) 
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

```python
bin_centers2 = bin_centers[:2*max_index]
count2       = counts[:2*max_index]

initial_guess = [peak_y, peak_x, 1.0]
popt, pcov = curve_fit(gaussian, bin_centers2, counts2, p0=initial_guess)

print(f"Fitted Mean: {popt[1]:.3f} MeV")
print(f"Fitted StdDev: {popt[2]:.3f} MeV")

# 6. Generate the fitted curve line
x_fit = np.linspace(0, 50, 1000) # Smooth x-axis for plotting the curve
y_fit = gaussian(x_fit, *popt)   # Unpack the fitted parameters into the function

# Plot the Gaussian fit
plt.plot(x_fit, y_fit, color='red', linewidth=2, label=f'Gaussian Fit ($\mu={popt[1]:.2f}$)')

# 7. Add labels and styling (incorporating your original code)
plt.title('Remnant excitation energy', fontsize=15)
plt.xlabel('Excitation Energy [MeV]', fontsize=12)
plt.xlim(0, 50)
plt.ylim(1e-4, 0.1) # Note: You had two ylims in your prompt, I kept the final one
plt.ylabel('Probability Density', fontsize=12) 
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()

# Show the plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Replace 'data.txt' with the actual path/name of your text file
file_path = 'E_miss_newsf3.txt'

# Read the text file. sep='\s+' handles spaces or tabs between columns
df = pd.read_csv(file_path, sep='\s+', header=None, names=['Column 1', 'Column 2','col3'], on_bad_lines='skip')



# Plot Histogram for the first column
#axes[0].hist(df['Column 1'], bins=2000, color='skyblue', edgecolor='black',density = "true")
#axes[0].set_title('Histogram of Column 1')
#axes[0].set_xlabel('E_miss')
#axes[0].set_xlim(0,30)
#axes[0].set_ylabel('Frequency')

# Create a figure with 1 row and 2 columns
fig, axes = plt.subplots(1, 3, figsize=(12, 5))

# Plot Histogram for the second col20mn
axes[0].hist(df['Column 2'], bins=1000, color='salmon', edgecolor='black',density = "true", histtype = "step")
axes[0].set_title('Histogram of Column 2')
axes[0].set_xlim(0,30)
axes[0].set_ylim(10**-3,1)
axes[0].set_yscale('log')
axes[0].set_xlabel('E_x')
axes[0].set_ylabel('Frequency')

axes[1].hist(df['col3'], bins=1000, color='salmon', edgecolor='black',density = "true", histtype = "step")
axes[1].set_title('Histogram of Column 2')
axes[1].set_xlim(0,30)
axes[1].set_ylim(10**-3,1)
axes[1].set_yscale('log')
axes[1].set_xlabel('E_x')
axes[1].set_ylabel('Frequency')

axes[2].hist(df['col3']-df['Column 2'], bins=1000, color='salmon', edgecolor='salmon',density = "true", histtype = "step")
axes[2].set_title('Histogram of Column 2')
axes[2].set_xlim(0,30)
axes[2].set_ylim(10**-3,1)
axes[2].set_yscale('log')
axes[2].set_xlabel('E_x')
axes[2].set_ylabel('Frequency')

```

```python
# Calculate the global minimum and maximum across both columns
min_val = min(df['col3'].min(), df['Column 2'].min())
max_val = max(df['col3'].max(), df['Column 2'].max())

# Create an array of bin edges to share between both histograms
# 1000 bins means 1001 bin edges
shared_bins = np.linspace(min_val, max_val, 1001)

# Pass 'shared_bins' to the bins parameter
plt.hist(df['col3'], bins=shared_bins, color='salmon', edgecolor='blue', density=True, histtype="step", label = "INCL + SF")
plt.hist(df['Column 2'], bins=shared_bins, color='salmon', edgecolor='salmon', density=True, histtype="step", label = "SF")

plt.title('Histogram of Column 2 and 3')
plt.xlim(-1, 25)
plt.ylim(10**-3, 1)
plt.yscale('log')
plt.xlabel('E_x')
plt.ylabel('Frequency')
plt.legend()
```
